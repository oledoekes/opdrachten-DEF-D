## Opdracht 2b: Opsplitsen taak 2: afleiding ontwerpeis

```python
# Bereken hier je benodigde kantelfrequentie
f =5
demping = 3
f_0 = f/np.sqrt(demping+1)
omega_0 = 2*np.pi*f_0
print(omega_0)
```

```python
# Voeg hier je plot in.
Frequentie_Array = np.array(np.linspace(0, 100, 100000))

def Verhouding(Frequentie, w_0):
    return 1/(1-((Frequentie**2)/(w_0**2)))

plt.figure()
plt.title("bodegrafiek")
plt.xlabel("Log Frequentie (log(Hz))")
plt.ylabel("log Verhouding Amplitudes (log(Am/Av))")
plt.plot(np.log(Frequentie_Array), np.log(np.abs(Verhouding(Frequentie_Array, omega_0))), 'r', label="bodegrafiek")
plt.legend()
plt.show()
```

```python
# Bereken hier de minimale Delta-x waarbij je aan de eis van de opdracht voldoet.
def delta_x(g, w_0):
    return g/(w_0**2)
delta_x_min = delta_x(9.81, omega_0)
print(delta_x_min)
```

## Opdracht 3: Plot de grafiek.

```python
# Voeg hier je plot in. Lees goed de opdracht wat er in de grafiek moet staan!
# Het is ook handig om je grafiek op te slaan, doe dat door met plt.savefig().
m = np.array(np.linspace(0, 1000, 1000))
C_min = ((m*9.81)/delta_x_min)*10e-3
C_2 = ((m*9.81)/(2*delta_x_min))*10e-3
C_5 = ((m*9.81)/(5*delta_x_min))*10e-3

plt.figure()
plt.title("parameterplot")
plt.xlabel("m(g)")
plt.ylabel("c(N/m)")
plt.plot(m, C_min, 'g', label="minimaal")
plt.plot(m, C_2, 'b', label="2 keer zo ruim")
plt.plot(m, C_5, 'r', label="5 keer zo ruim")
plt.legend()
plt.show()
```

```python
# Voeg hier je plot in.
from scipy.optimize import curve_fit
Data_Av_it0 = np.array([0.6, 1.68, 2.69, 3.54, 6.23, 8.89, 3.77])
Data_Am_it0 = np.array([0.36, 1.27, 2.30, 1.8, 4.53, 1.82, 0.27])
Data_f_it0 = np.array([1.6, 2.7, 3.95, 5.25, 7.1, 7.95, 9.6])
u_Av_it0 = np.array([0.05, 0.03, 0.03, 0.06, 0.15, 0.1, 0.04])
u_Am_it0 = np.array([0.05, 0.03, 0.03, 0.06, 0.15, 0.1, 0.04])
u_f_it0 = np.array([0.1, 0.2, 0.3, 0.1, 0.2, 0.3, 0.4])

Verhouding_Meting_it0 = Data_Am_it0/Data_Av_it0
u_A_it0 = (u_Av_it0+ u_Am_it0)/2

Fit_xas = np.array(np.logspace(0, 1.5, 100))
def Fit_Verhouding(Frequentie, w_0):
    return 1/(1-(((Frequentie*2*np.pi)**2)/(w_0**2)))
val, cov = curve_fit(Fit_Verhouding, np.log(Data_f_it0), np.log(Verhouding_Meting_it0), p0=10)
print(val) # curve fit berekend te veel punten tussen 0 en 1 en dat zorgt waarschijnlijk voor de gehalicuneerde waarde van omega0
plt.figure()
plt.title("iteratie_0")
plt.xlabel("log van de frequentie")
plt.ylabel("log van de verhouding amplitudes")
#plt.ylim(15, -15) deze bleek niet nodig te zijn
plt.plot(Data_f_it0, np.abs(Verhouding_Meting_it0), 'b.', label="meting iteratie0")
plt.plot(Fit_xas, np.abs(Fit_Verhouding(Fit_xas, 8)), 'g', label="Transfer Curve")
plt.xscale("log")
plt.yscale("log")
# plt.plot(np.log(Fit_xas), np.log(np.abs(Fit_Verhouding(Fit_xas, val))), 'r', label="curve_fit") 
#plt.errorbar(u_f_it0, u_A_it0, np.log(Data_f_it0), np.log(np.abs(Verhouding_Meting_it0)), 'r.', label="errorbars") weet niet hoe ik een errorbar moet verwerken op een logaritmische schaal
plt.legend()
plt.savefig("Iteratie0")
plt.show()
```
